## 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

## 2: Load & Prepare Data

```python
df = pd.read_csv("../data/indoorAir.csv")

df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s')
df = df.sort_values(by='timestamp')
df.set_index('timestamp', inplace=True)
```

## 3. Time Series Visualization

```python
plt.figure(figsize=(10,4))
plt.plot(df['scd41_co2'])
plt.title("CO2 Time Series")
plt.show()
```

## 4. Heatmap

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.select_dtypes(include=['number']).corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

## 5. Feature Selection (DATA-DRIVEN)(Hybrid Method)

```python
# Random Forest selected features (your list)
rf_features = [
    'scd41_humidity',
    'bme688_temperature',
    'scd41_temperature',
    'bme688_humidity',
    'svm41_temperature',
    'ens160_tvoc',
    'ens160_eco2',
    'svm41_humidity',
    'sfa30_temperature',
    'bme688_pressure'
]

# Target column
target_feature = ['scd41_co2']

# Combine
selected_features = rf_features + target_feature

print("Final Selected Features:", selected_features)

# Apply selection
df = df[selected_features]
```

## 6. Resample Data

```python
df = df.resample('15min').mean()
print("Total rows:", len(df))
```

## 7. Handle Missing Values

```python
# Handle missing values (final clean version)
df = df.interpolate(method='time')   # fill based on time
df = df.ffill()                      # forward fill remaining
df = df.bfill()                      # backward fill edge cases

# Final safety check
df = df.dropna()

print("Remaining NaNs:", df.isna().sum().sum())
```

## 8. Train / Validation / Test Split

```python
# Fixed Number
#data = df.values

#train_size = 17317
#val_size = 3710
#test_size = 3712

#train_data = data[:train_size]

#val_data = data[train_size:train_size + val_size]

#test_data = data[train_size + val_size:train_size + val_size + test_size]

#print("Train shape:", train_data.shape)
#print("Validation shape:", val_data.shape)
#print("Test shape:", test_data.shape)

# Automatically from total rows
data = df.values

total_rows = len(data)

train_size = int(0.7 * total_rows)
val_size = int(0.15 * total_rows)
test_size = total_rows - train_size - val_size

train_data = data[:train_size]
val_data = data[train_size:train_size + val_size]
test_data = data[train_size + val_size:]

print("Train rows:", len(train_data))
print("Validation rows:", len(val_data))
```

## 9. Scaling (Normalization)

```python
#used sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

train_data = scaler.fit_transform(train_data)
val_data = scaler.transform(val_data)
test_data = scaler.transform(test_data)
```

## 10. Create Sequences

```python
def create_sequences(data, seq_length=24):
    X, y = [], []

    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length][0])  # CO2 target

    return np.array(X), np.array(y)

X_train, y_train = create_sequences(train_data)
X_val, y_val = create_sequences(val_data)
X_test, y_test = create_sequences(test_data)
```

## 11. Convert to Tensor

```python
#import torch

X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)

X_val = torch.tensor(X_val, dtype=torch.float32)
y_val = torch.tensor(y_val, dtype=torch.float32)

X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)
```

## 12. Create DataLoaders

```python
#from torch.utils.data import DataLoader, TensorDataset

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=32, shuffle=True)
val_loader = DataLoader(TensorDataset(X_val, y_val), batch_size=32, shuffle=False)
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=32, shuffle=False)
```

## 13. LSTM+ CNN Model

```python
#import torch.nn as nn

class CNN_LSTM_Model(nn.Module):
    def __init__(self, input_size, hidden_size=64, num_layers=2):
        super().__init__()
        
        self.conv1 = nn.Conv1d(input_size, 128, kernel_size=5, padding=2)
        self.conv2 = nn.Conv1d(64, 128, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.3)
        
        self.lstm = nn.LSTM(input_size=128, hidden_size=64, batch_first=True)

        # Fully connected
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        # x shape: (batch, seq_len, features)

        # 🔄 Change to (batch, features, seq_len) for CNN
        x = x.permute(0, 2, 1)

        # CNN
        x = self.conv1(x)
        x = self.relu(x)
        

        # 🔄 Back to (batch, seq_len, features) for LSTM
        x = x.permute(0, 2, 1)

        # LSTM
        out, _ = self.lstm(x)

        # Take last time step
        out = out[:, -1, :]

        return self.fc(out)
```

## 14. Initialize Model

```python
model = CNN_LSTM_Model(input_size=X_train.shape[2])

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

## 15. Training

```python
for epoch in range(10):
    model.train()
    total_loss = 0

    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()

        outputs = model(X_batch)
        loss = criterion(outputs.squeeze(), y_batch)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss:.4f}")
```

## 16. Evaluation

```python
model.eval()
predictions, actuals = [], []

with torch.no_grad():
    for X_batch, y_batch in test_loader:
        outputs = model(X_batch)
        predictions.extend(outputs.squeeze().numpy())
        actuals.extend(y_batch.numpy())

from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

mse = mean_squared_error(actuals, predictions)
mae = mean_absolute_error(actuals, predictions)
rmse = np.sqrt(mse)

print("MSE:", mse)
print("MAE:", mae)
print("RMSE:", rmse)
```

## 17. Visualization

```python
plt.figure(figsize=(10,4))
plt.plot(actuals[:200], label="Actual")
plt.plot(predictions[:200], label="Predicted")
plt.legend()
plt.title("Actual vs Predicted CO2")
plt.show()
```
