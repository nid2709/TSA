```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ruptures as rpt

from scipy.cluster.hierarchy import dendrogram, linkage
from pandas.plotting import lag_plot, autocorrelation_plot
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller, ccf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

## 1. Load Dataset

```python
df = pd.read_csv("../data/indoorAir.csv")

df.head()
```

## 2. Understand Structure of Data

```python
df.shape
df.info()
df.describe()
```

## 3. Handle Time Column

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s') 
df = df.sort_values(by='timestamp')
df.set_index('timestamp', inplace=True)


df.head()
```

## 4. Time Series Visualization

```python
# 1. CO2 (best starting point)
plt.figure(figsize=(12,5))
plt.plot(df['scd41_co2'], label='CO2 (SCD41)')
plt.title('CO2 Time Series')
plt.xlabel('Time')
plt.ylabel('CO2 (ppm)')
plt.legend()
plt.show()
```

### -> Plot Multiple Important Features

```python
# Multiple key air quality variables
df[['scd41_co2', 'ens160_aqi', 'ens160_tvoc']].plot(figsize=(12,5))

plt.title('Air Quality Parameters Over Time')
plt.xlabel('Time')
plt.ylabel('Values')
plt.show()
```

### -> Temperature Comparison (Different Sensors)

```python
df[['scd41_temperature', 'svm41_temperature', 'bme688_temperature']].plot(figsize=(12,5))

plt.title('Temperature Comparison Across Sensors')
plt.xlabel('Time')
plt.ylabel('Temperature (°C)')
plt.show()
```

## 5. Trend and Seasonality Analysis

```python
# Select variable
ts = df['scd41_co2']

# Resample
ts = ts.resample('1h').mean()

# Handle missing values (IMPORTANT)
ts = ts.interpolate()

# Optional: check
print("Missing values:", ts.isna().sum())

# Decomposition
decomposition = seasonal_decompose(ts, model='additive')

# Plot
decomposition.plot()
plt.show()
```

## 6. Correlation Analysis (Heatmap)

```python
# Select only numeric columns (ignore id, station info)
corr = df.select_dtypes(include=['float64', 'int64']).corr()

# Plot heatmap
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Correlation Heatmap')
plt.show()
```

## 7. Autocorrelation Analysis (ACF/PACF)

```python
# Select main variable
ts = df['scd41_co2']

# Optional: resample (recommended)
ts = ts.resample('1h').mean().dropna()

# Plot ACF
plt.figure(figsize=(12,4))
plot_acf(ts, lags=50)
plt.title('Autocorrelation Function (ACF)')
plt.show()

# Plot PACF
plt.figure(figsize=(12,4))
plot_pacf(ts, lags=50)
plt.title('Partial Autocorrelation Function (PACF)')
plt.show()
```

## 8. Cross-Correlation Analysis (Between Variables)

```python
# Select two variables
x = df['scd41_co2']
y = df['ens160_aqi']

# Resample (important)
x = x.resample('1h').mean().dropna()
y = y.resample('1h').mean().dropna()

# Align both series
df_cc = x.to_frame('x').join(y.to_frame('y')).dropna()

# Compute cross-correlation
ccf_values = ccf(df_cc['x'], df_cc['y'])

# Plot
plt.figure(figsize=(10,4))
plt.plot(ccf_values[:50])
plt.title('Cross-Correlation (CO2 vs AQI)')
plt.xlabel('Lag')
plt.ylabel('Correlation')
plt.show()
```

## 9. Stationarity Testing (ADF Test)

```python
# Select variable
ts = df['scd41_co2']

# Resample (recommended)
ts = ts.resample('1h').mean().dropna()

# ADF Test
result = adfuller(ts)

# Output results
print("ADF Statistic:", result[0])
print("p-value:", result[1])

# Critical values
for key, value in result[4].items():
    print(f'Critical Value ({key}): {value}')
```

## 10. Rolling Statistics (Moving Average and Variance)

```python
# Select variable
ts = df['scd41_co2']

# Resample (recommended)
ts = ts.resample('1h').mean().dropna()

# Rolling calculations
rolling_mean = ts.rolling(window=24).mean()   # 24 hours
rolling_std = ts.rolling(window=24).std()

# Plot
plt.figure(figsize=(12,5))
plt.plot(ts, label='Original')
plt.plot(rolling_mean, label='Rolling Mean')
plt.plot(rolling_std, label='Rolling Std')

plt.title('Rolling Statistics (CO2)')
plt.xlabel('Time')
plt.legend()
plt.show()
```

## 11. Heatmap over Time (Hour of Day/ Day of Week)

```python
# Create time-based features
df['hour'] = df.index.hour
df['day_of_week'] = df.index.day_name()

# Pivot table (average CO2)
pivot = df.pivot_table(
    values='scd41_co2',
    index='day_of_week',
    columns='hour',
    aggfunc='mean'
)

# Reorder days properly
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
pivot = pivot.reindex(days_order)

# Plot heatmap
plt.figure(figsize=(12,6))
sns.heatmap(pivot, cmap='coolwarm')

plt.title('CO2 Levels by Day of Week and Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.show()
```

## 12. Outlier Detection

### -> Boxplot Method

```python
# Boxplot for CO2
plt.figure(figsize=(8,4))
sns.boxplot(x=df['scd41_co2'])

plt.title('Boxplot for CO2 (Outlier Detection)')
plt.show()
```

### -> QR Method – Detect Outliers

```python
# IQR method
Q1 = df['scd41_co2'].quantile(0.25)
Q3 = df['scd41_co2'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df['scd41_co2'] < lower_bound) | (df['scd41_co2'] > upper_bound)]

print("Number of outliers:", len(outliers))
```

## 14. Change Point Detection

```python
# Select variable
# ts = df['scd41_co2'].dropna().values

# Model (l2 = mean shift detection)
# model = "l2"

# Fit model
# algo = rpt.Pelt(model=model).fit(ts)

# Detect change points (penalty can be tuned)
# change_points = algo.predict(pen=10)

# Plot
# plt.figure(figsize=(12,5))
# plt.plot(ts, label='CO2')

# Mark change points
# for cp in change_points:
#     plt.axvline(cp, color='red', linestyle='--')

# plt.title('Change Point Detection (CO2)')
# plt.legend()
# plt.show()

# Select variable
ts = df['scd41_co2'].dropna().values

# Reduce size (VERY important for big datasets)
ts = ts[:5000]

# Model
model = "l2"

# Use faster algorithm instead of PELT
algo = rpt.Binseg(model=model).fit(ts)

# Detect fixed number of change points (fast)
change_points = algo.predict(n_bkps=5)

# Plot
plt.figure(figsize=(12,5))
plt.plot(ts, label='CO2')

# Mark change points
for cp in change_points:
    plt.axvline(cp, color='red', linestyle='--')

plt.title('Change Point Detection (CO2)')
plt.legend()
plt.show()
```

## 15. Principle Component Analysis (PCA)

```python
# Drop obvious non-useful columns
features = df.drop(columns=['id', 'station_id', 'station_name'], errors='ignore')

# Keep ONLY numeric columns
features = features.select_dtypes(include=[np.number])

# Check
print("After selecting numeric:", features.shape)

# Handle missing values
features = features.interpolate()   # now safe
features = features.dropna()

print("After cleaning:", features.shape)

# Standardize
scaler = StandardScaler()
scaled_data = scaler.fit_transform(features)

# PCA
pca = PCA()
pca_data = pca.fit_transform(scaled_data)

# Explained variance
explained_variance = pca.explained_variance_ratio_

print(explained_variance)
```

## 16. Feature Importance Exploring

```python
# # Target
# y = df['scd41_co2']

# # Features
# X = df.drop(columns=['id', 'station_id', 'station_name', 'scd41_co2'], errors='ignore')

# # Keep only numeric
# X = X.select_dtypes(include=[np.number])

# # Handle missing values (key fix)
# X = X.interpolate()
# y = y.interpolate()

# # Combine and clean remaining NaNs
# data = pd.concat([X, y], axis=1).dropna()

# X = data.drop(columns=['scd41_co2'])
# y = data['scd41_co2']

# print("Final shape:", X.shape)

# # Train model
# model = RandomForestRegressor(n_estimators=100, random_state=42)
# model.fit(X, y)

# Feature importance
# importance = pd.Series(model.feature_importances_, index=X.columns)
# importance = importance.sort_values(ascending=False)

# # Plot
# plt.figure(figsize=(10,6))
# importance.head(10).plot(kind='bar')

# plt.title('Top 10 Feature Importance')
# plt.ylabel('Importance Score')
# plt.show()


# Target
y = df['scd41_co2']

# Features
X = df.drop(columns=['id', 'station_id', 'station_name', 'scd41_co2'], errors='ignore')

# Keep only numeric
X = X.select_dtypes(include=[np.number])

# Handle missing values
X = X.interpolate()
y = y.interpolate()

# Combine and clean remaining NaNs
data = pd.concat([X, y], axis=1).dropna()

X = data.drop(columns=['scd41_co2'])
y = data['scd41_co2']

print("Final shape:", X.shape)


X = X.iloc[:5000]
y = y.iloc[:5000]

# Train model (OPTIMIZED)
model = RandomForestRegressor(
    n_estimators=50,      # reduced from 100
    max_depth=10,         # limits tree size
    n_jobs=-1,            # uses all CPU cores
    random_state=42
)

model.fit(X, y)

# Feature importance
importance = pd.Series(model.feature_importances_, index=X.columns)
importance = importance.sort_values(ascending=False)

# Plot
plt.figure(figsize=(10,6))
importance.head(10).plot(kind='bar')

plt.title('Top 10 Feature Importance')
plt.ylabel('Importance Score')
plt.show()
```

## 17. Clustering Analysis

### -> K-Means Clustering 

```python
features = df.drop(columns=['id', 'station_id', 'station_name'], errors='ignore')

features = features.select_dtypes(include=['number'])

features = features.fillna(features.mean())

# If still empty (safety check)
if features.shape[0] == 0:
    raise ValueError("Dataset is empty after preprocessing. Check missing values.")

scaler = StandardScaler()
scaled_data = scaler.fit_transform(features)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(scaled_data)

# Add cluster labels
features = features.copy()
features['cluster'] = clusters

if 'scd41_co2' in features.columns and 'ens160_aqi' in features.columns:
    plt.figure(figsize=(8,5))
    plt.scatter(
        features['scd41_co2'],
        features['ens160_aqi'],
        c=clusters,
        cmap='viridis'
    )

    plt.title('K-Means Clustering (CO2 vs AQI)')
    plt.xlabel('CO2')
    plt.ylabel('AQI')
    plt.show()
else:
    print("Required columns (scd41_co2, ens160_aqi) not found for plotting.")
```

### -> Hierarchical Clustering

```python
# STEP 1: reduce dataset size
scaled_small = scaled_data[:5000]

# STEP 2: hierarchical clustering
linked = linkage(scaled_small, method='ward')

# STEP 3: plot
plt.figure(figsize=(10,5))
dendrogram(linked, truncate_mode='level', p=5)

plt.title('Hierarchical Clustering Dendrogram')
plt.show()
```
