## 1: Import Libraries

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.preprocessing import MinMaxScaler
```

## 1. Load Dataset

```python
df = pd.read_csv("../data/indoorAir.csv")
```

## 2. Convert Timestamp & Set Index

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s')
df = df.sort_values(by='timestamp')
df.set_index('timestamp', inplace=True)
```

## 3. Select Relevant Features

```python
features = [
    'scd41_co2',
    'scd41_temperature',
    'scd41_humidity',
    'ens160_tvoc',
    'ens160_aqi'
]

df = df[features]
```

## 4. Resample Data (Hourly)

```python
df = df.resample('1h').mean()
```

## 5. Handle Missing Values

```python
df = df.interpolate(method='time')
df = df.ffill()
```

## 6. Normalize Data (Scaling)

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns, index=df.index)
```

## 7. Create Sequences (For LSTM)

```python
def create_sequences(data, seq_length=24):
    X = []
    y = []

    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length][0])  # CO2 as target

    return np.array(X), np.array(y)

X, y = create_sequences(scaled_df.values, seq_length=24)
#From now on 
#X → input sequences (past 24 hours data)
#y → target (next CO₂ value)
```

## 8. Train / Validation / Test Split

```python
# Define split sizes
train_size = int(0.7 * len(X))
val_size = int(0.15 * len(X))

# Split sequentially (no shuffle!)
X_train = X[:train_size]
y_train = y[:train_size]

X_val = X[train_size:train_size + val_size]
y_val = y[train_size:train_size + val_size]

X_test = X[train_size + val_size:]
y_test = y[train_size + val_size:]
```

## 9: Convert to Tensors (PyTorch Requirement)

```python
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)

X_val = torch.tensor(X_val, dtype=torch.float32)
y_val = torch.tensor(y_val, dtype=torch.float32)

X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)
```

## 10: Create Dataset & Dataloader

```python
train_dataset = TensorDataset(X_train, y_train)
val_dataset = TensorDataset(X_val, y_val)
test_dataset = TensorDataset(X_test, y_test)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

## 11. Define LSTM Model

```python
class LSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers):
        super(LSTMModel, self).__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,   # number of features
            hidden_size=hidden_size, # neurons
            num_layers=num_layers,   # stacked layers
            batch_first=True
        )

        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = out[:, -1, :]   # last time step
        out = self.fc(out)
        return out
```

## 12: Initialize Model

```python
input_size = X.shape[2]   # 5 features
hidden_size = 64
num_layers = 2

model = LSTMModel(input_size, hidden_size, num_layers)
```

## 13: Loss Function & Optimizer

```python
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

## 14: Training Loop

```python
num_epochs = 10 #Model will see the entire dataset 10 times

for epoch in range(num_epochs): #Each loop = one full training cycle
    model.train()
    total_loss = 0 #To track total error in one epoch

    for X_batch, y_batch in train_loader: #Instead of whole data, we train in small batches (32 samples)
        optimizer.zero_grad() #Clears old gradients and Prevents accumulation

        outputs = model(X_batch) # Model makes prediction
        loss = criterion(outputs.squeeze(), y_batch) # Compare predicted value and actual value

        loss.backward() # Which weights cause error and how to correct them
        optimizer.step() #Adjust model parameter to reduce error

        total_loss += loss.item()#Stop error value for monitoring

    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {total_loss:.4f}") # also shows how model is improving
```

## 15. Validation

```python
model.eval()
val_loss = 0

with torch.no_grad():
    for X_batch, y_batch in val_loader:
        outputs = model(X_batch)
        loss = criterion(outputs.squeeze(), y_batch)
        val_loss += loss.item()

print("Validation Loss:", val_loss)
```

## 16: Testing + Metrics

```python
model.eval()
predictions = []
actuals = []

with torch.no_grad():
    for X_batch, y_batch in test_loader:
        outputs = model(X_batch)
        predictions.extend(outputs.squeeze().numpy())
        actuals.extend(y_batch.numpy())

mse = mean_squared_error(actuals, predictions)
mae = mean_absolute_error(actuals, predictions)

print("MSE:", mse)
print("MAE:", mae)
```
