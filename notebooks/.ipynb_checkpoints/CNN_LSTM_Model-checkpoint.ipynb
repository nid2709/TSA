```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM, Dense, Dropout
```

## 1. Load Data

```python
df = pd.read_csv("../data/indoorAir.csv")
df.head()
```

```python
print(df.shape)
print(df.info())
print(df.describe())
```

## 2. EDA

```python
# Missing values
missing = df.isnull().sum().sort_values(ascending=False)
print(missing)
```

```python
# Drop columns with >50% missing
df = df.dropna(axis=1, thresh=len(df)*0.5)
print("Columns kept:", df.columns.tolist())
```

```python
# Timestamp conversion
df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s')
df = df.sort_values('timestamp').reset_index(drop=True)
df.head()
```

```python
# Correlation Heatmap
corr = df.select_dtypes(include=['float64', 'int64']).corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# CO2 Time Series
plt.figure(figsize=(14, 4))
plt.plot(df['timestamp'], df['scd41_co2'], lw=0.6, color='teal')
plt.title('CO2 over Time')
plt.xlabel('Time')
plt.ylabel('CO2 (ppm)')
plt.tight_layout()
plt.show()
```

```python
# Boxplot — Outlier Detection
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['scd41_co2'], color='lightblue')
plt.title('CO2 Boxplot — Outlier Detection')
plt.show()
```

```python
# Hourly Seasonality
df['hour'] = df['timestamp'].dt.hour
hourly = df.groupby('hour')['scd41_co2'].mean()
plt.figure(figsize=(10, 4))
plt.plot(hourly.index, hourly.values, 'o-', color='darkorange')
plt.title('Mean CO2 by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Mean CO2 (ppm)')
plt.xticks(range(0, 24))
plt.tight_layout()
plt.show()
```

```python
# Distribution of CO2
plt.figure(figsize=(8, 4))
sns.histplot(df['scd41_co2'], kde=True, color='steelblue', bins=60)
plt.title('CO2 Distribution')
plt.xlabel('CO2 (ppm)')
plt.show()
```

```python
# Rolling Mean & Std
co2 = df['scd41_co2']
plt.figure(figsize=(14, 4))
plt.plot(co2.values[-3000:], lw=0.5, alpha=0.5, label='Raw', color='gray')
plt.plot(co2.rolling(60).mean().values[-3000:], label='60-pt rolling mean', color='blue')
plt.title('CO2 Rolling Mean')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Preprocessing

```python
# Use one station for clean modelling
df = df[df['station_name'] == 'station_room_233'].copy()

# Drop non-feature columns
features = df.drop(['id', 'station_id', 'station_name', 'timestamp', 'hour'], axis=1)

# Forward-fill any remaining NaNs
features = features.ffill().dropna()
print(features.shape)
print(features.columns.tolist())
```

```python
# Outlier clipping (Winsorization)
for col in features.columns:
    q01, q99 = features[col].quantile(0.01), features[col].quantile(0.99)
    features[col] = features[col].clip(q01, q99)
```

```python
# Scaling
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(features)
```

```python
# Sequence creation — window=60 steps, predict CO2
def create_sequences(data, target_index, window=60):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i+window])
        y.append(data[i+window][target_index])
    return np.array(X), np.array(y)

target_index = list(features.columns).index('scd41_co2')
X, y = create_sequences(scaled_data[:50000], target_index)
print("X shape:", X.shape, " | y shape:", y.shape)
```

```python
# Train / Test split (chronological — NO shuffle)
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

## 4. CNN-LSTM Model

```python
model = Sequential()

# CNN — extracts local short-term patterns
model.add(Conv1D(filters=64, kernel_size=3, activation='relu',
                 input_shape=(X.shape[1], X.shape[2])))
model.add(MaxPooling1D(pool_size=2))

# LSTM — captures long-range temporal dependencies
model.add(LSTM(100, return_sequences=False))

# Dropout — prevents overfitting
model.add(Dropout(0.2))

# Output layer
model.add(Dense(1))

model.compile(optimizer='adam', loss='mse')
model.summary()
```

## 5. Training

```python
history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=64,
    validation_data=(X_test, y_test)
)
```

```python
# Training curves
plt.figure(figsize=(10, 4))
plt.plot(history.history['loss'],     label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Training & Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE')
plt.legend()
plt.show()
```

## 6. Evaluation

```python
loss = model.evaluate(X_test, y_test)
print("Test MSE Loss:", loss)
```

```python
predictions = model.predict(X_test)

# Inverse transform CO2 only
def inv_co2(vals, scaler, idx, n_feat):
    dummy = np.zeros((len(vals), n_feat))
    dummy[:, idx] = vals.flatten()
    return scaler.inverse_transform(dummy)[:, idx]

n_feat = features.shape[1]
y_actual = inv_co2(y_test,       scaler, target_index, n_feat)
y_pred   = inv_co2(predictions,  scaler, target_index, n_feat)

rmse = np.sqrt(mean_squared_error(y_actual, y_pred))
mae  = mean_absolute_error(y_actual, y_pred)
r2   = r2_score(y_actual, y_pred)

print(f"RMSE : {rmse:.4f} ppm")
print(f"MAE  : {mae:.4f} ppm")
print(f"R²   : {r2:.4f}")
```

## 7. Results

```python
# Actual vs Predicted
plt.figure(figsize=(14, 4))
plt.plot(y_actual[:200], label='Actual',    color='steelblue')
plt.plot(y_pred[:200],   label='Predicted', color='tomato', linestyle='--')
plt.title('Actual vs Predicted CO2')
plt.xlabel('Time Steps')
plt.ylabel('CO2 (ppm)')
plt.legend()
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(6, 6))
plt.scatter(y_actual[:2000], y_pred[:2000], alpha=0.3, s=8, color='purple')
lims = [min(y_actual.min(), y_pred.min()), max(y_actual.max(), y_pred.max())]
plt.plot(lims, lims, 'r--', lw=1.5, label='Perfect')
plt.xlabel('Actual CO2 (ppm)')
plt.ylabel('Predicted CO2 (ppm)')
plt.title(f'Scatter  |  R² = {r2:.4f}')
plt.legend()
plt.show()
```
